```python

import pandas as pd

df = pd.read_csv(r'c:\Users\carso\PyCharmMiscProject\training_log.csv')

df["epoch_number"] = range(1, len(df) + 1)

df = df.drop(columns="epochs")
print(df.head())
```

```python
acc = df[" accuracy"]
loss = df[" loss"]
val_acc = df[" val_accuracy"]
val_loss = df[" val_loss"]
epoch_number = df["epoch_number"]

if loss.shape == acc.shape == val_acc.shape == val_loss.shape == epoch_number.shape: # make sure none has false data
    print("good")
```

```python
numbers = []

for i in range(1, len(df) + 1):
    numbers.append(i)

df["epoch_number"] = numbers

df.head()
```

```python
#convert each to precent


# removes hidden spaces from column names
df.columns = df.columns.str.strip()



# convert all model metric columns to percent
df["accuracy_percent"] = (df["accuracy"] * 100).round(2)
df["loss_percent"] = (df["loss"] * 100).round(2)
df["val_accuracy_percent"] = (df["val_accuracy"] * 100).round(2)
df["val_loss_percent"] = (df["val_loss"] * 100).round(2)

df.head()
```

```python
plt.figsize=(10,6)

plt.plot(df["epoch_number"], df["accuracy_percent"], label="Accuracy %")
plt.plot(df["epoch_number"], df["loss_percent"], label="Loss %")
plt.plot(df["epoch_number"], df["val_accuracy_percent"], label="Validation Accuracy %")
plt.plot(df["epoch_number"], df["val_loss_percent"], label="Validation Loss %")

plt.xlabel("Epoch Number")
plt.ylabel("Percent")
plt.title("Training Metrics Over Epochs")
plt.legend()
plt.grid(True)

plt.show()
```

* Training Accuracy increased to almost 100 meaning a good fit for trianing data
* training loss decreased over epoch
* validation had little imporvement
* validation loss increased substantually likely meaning the model was overfitting data

* Overall: 
* The model learned training data well but did not imoprove on unseen data after early epochs.
